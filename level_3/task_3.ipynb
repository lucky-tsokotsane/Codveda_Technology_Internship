**REMOVING DODGY IMAGES**

These images may be corrupted or mislead the model.

```python
base_dir = "sample_data/chest_xray"
test_dir = "test"
train_dir = "train"
val_dir = "val"

directories = [test_dir, train_dir, val_dir]

for dir in directories:
  sub_dirs = os.listdir(os.path.join(base_dir, dir))

  for sub_dir in sub_dirs:
    for picture in os.listdir(os.path.join(base_dir, dir, sub_dir)):
      if round(os.path.getsize(os.path.join(base_dir, dir, sub_dir, picture)) / 1000) < 10:
        os.remove(os.path.join(base_dir, dir, sub_dir, picture))
      else:
        continue

print("Dodgy images removed.")
```
